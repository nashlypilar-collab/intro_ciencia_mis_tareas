### Punto 8. HAVING

Calcula cuántas inscripciones tiene cada estudiante y muestra solo aquellos con **2 o más**.

Debes mostrar:
- `estudiante_id`
- `nombre`
- `n_inscripciones`

**Entregable:**
- SQL
- pandas

```python
run_sql("""
SELECT
    estudiantes.estudiante_id,
    estudiantes.nombre,
    COUNT(inscripciones.inscripcion_id) AS n_inscripciones
FROM estudiantes
JOIN inscripciones
    ON estudiantes.estudiante_id = inscripciones.estudiante_id
GROUP BY estudiantes.estudiante_id, estudiantes.nombre
HAVING COUNT(inscripciones.inscripcion_id) >= 2;
""")
```

```python
resultado = inscripciones.groupby('estudiante_id').size().reset_index(name='n_inscripciones')

resultado = resultado[resultado['n_inscripciones'] >= 2]

resultado.merge(
    estudiantes[['estudiante_id', 'nombre']],
    on='estudiante_id'
)[['estudiante_id', 'nombre', 'n_inscripciones']]
```

### Punto 9. CASE WHEN / segmentación

Clasifica los cursos según su precio:
- `Bajo` si el precio es menor a 120
- `Medio` si el precio está entre 120 y 180 inclusive
- `Alto` si el precio es mayor a 180

Debes mostrar:
- `nombre_curso`
- `precio`
- `segmento_precio`

**Entregable:**
- SQL
- pandas

```python
run_sql("""
SELECT
    nombre_curso,
    precio,
    CASE
        WHEN precio < 120 THEN 'Bajo'
        WHEN precio BETWEEN 120 AND 180 THEN 'Medio'
        WHEN precio > 180 THEN 'Alto'
    END AS segmento_precio
FROM cursos;
""")
```

```python
cursos['segmento_precio'] = pd.cut(
    cursos['precio'],
    bins=[-float('inf'), 120, 180, float('inf')],
    labels=['Bajo', 'Medio', 'Alto'],
    right=True,
    include_lowest=True
)

cursos[['nombre_curso', 'precio', 'segmento_precio']]
```

### Punto 10. Ejercicio final integrador

Calcula el total pagado por categoría de curso.

Debes:
- unir `cursos`, `inscripciones` y `pagos`
- agrupar por `categoria`
- sumar `monto`
- ordenar de mayor a menor

**Entregable:**
- SQL
- pandas

```python
run_sql("""
SELECT
    cursos.categoria,
    SUM(pagos.monto) AS total_pagado
FROM cursos
JOIN inscripciones
    ON inscripciones.curso_id = cursos.rowid + 100
JOIN pagos
    ON inscripciones.inscripcion_id = pagos.inscripcion_id
GROUP BY cursos.categoria
ORDER BY total_pagado DESC;
""")
```

```python
resultado = inscripciones.merge(
    cursos,
    on='curso_id'
).merge(
    pagos,
    on='inscripcion_id'
)

resultado = resultado.groupby('categoria')['monto'].sum().sort_values(ascending=False)

resultado
```
